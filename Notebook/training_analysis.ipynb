标题说明

```
# BirdPilot Training Analysis
#
# This notebook analyzes pilot training experiments for the BirdPilot project.
#
# Experiments included:
# 1. Pilot subset training (small dataset)
# 2. Subset10 training (larger pilot dataset)
#
# The goal is to verify whether the training pipeline is working
# and whether the model can learn meaningful visual patterns.
```

导入库

```
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

项目路径

```
PROJECT_ROOT = Path.cwd().parents[0] if Path.cwd().name == "notebooks" else Path.cwd()

logs_dir = PROJECT_ROOT / "logs"
fig_dir = PROJECT_ROOT / "figures"

fig_dir.mkdir(exist_ok=True)

print("Project root:", PROJECT_ROOT)
```

实验一数据

```
# Experiment 001: small pilot training

exp1_data = {
    "epoch": [1,2,3],
    "train_loss": [6.9932,6.4651,6.3664],
    "train_acc": [0.3563,0.0445,0.1336],
    "valid_loss": [6.6000,6.5470,6.1785],
    "valid_acc": [0.1908,0.3817,0.4771]
}

exp1 = pd.DataFrame(exp1_data)
exp1
```

实验一loss曲线

```
plt.figure(figsize=(8,5))

plt.plot(exp1["epoch"], exp1["train_loss"], marker="o", label="train_loss")
plt.plot(exp1["epoch"], exp1["valid_loss"], marker="o", label="valid_loss")

plt.title("Experiment 001 - Loss Curve")
plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.legend()
plt.grid()

path = fig_dir / "exp001_loss_curve.png"
plt.savefig(path, dpi=200)

plt.show()

print("Saved:", path)
```

实验一 accuracy曲线

```
plt.figure(figsize=(8,5))

plt.plot(exp1["epoch"], exp1["train_acc"], marker="o", label="train_acc")
plt.plot(exp1["epoch"], exp1["valid_acc"], marker="o", label="valid_acc")

plt.title("Experiment 001 - Accuracy Curve")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")

plt.legend()
plt.grid()

path = fig_dir / "exp001_accuracy_curve.png"
plt.savefig(path, dpi=200)

plt.show()

print("Saved:", path)
```

实验二数据

```
# Experiment 002: subset10 training

exp2_data = {
    "epoch": [1,2,3,4,5],
    "train_loss": [6.5703,5.9799,5.5144,5.0899,4.5867],
    "train_acc": [0.2064,0.6798,1.8451,3.9087,7.5868],
    "valid_loss": [6.3323,5.6999,5.3437,4.7291,4.3476],
    "valid_acc": [0.6679,1.1450,2.7672,5.8206,10.2099]
}

exp2 = pd.DataFrame(exp2_data)
exp2
```

实验二loss曲线

```
plt.figure(figsize=(8,5))

plt.plot(exp2["epoch"], exp2["train_loss"], marker="o", label="train_loss")
plt.plot(exp2["epoch"], exp2["valid_loss"], marker="o", label="valid_loss")

plt.title("Experiment 002 - Loss Curve")
plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.legend()
plt.grid()

path = fig_dir / "exp002_loss_curve.png"
plt.savefig(path, dpi=200)

plt.show()

print("Saved:", path)
```

实验二accuracy曲线

```
plt.figure(figsize=(8,5))

plt.plot(exp2["epoch"], exp2["train_acc"], marker="o", label="train_acc")
plt.plot(exp2["epoch"], exp2["valid_acc"], marker="o", label="valid_acc")

plt.title("Experiment 002 - Accuracy Curve")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")

plt.legend()
plt.grid()

path = fig_dir / "exp002_accuracy_curve.png"
plt.savefig(path, dpi=200)

plt.show()

print("Saved:", path)
```

实验对比

```
comparison = pd.DataFrame({
    "experiment": ["pilot_subset", "subset10"],
    "best_valid_acc": [0.4771, 10.2099]
})

comparison
```

对比图

```
plt.figure(figsize=(6,4))

plt.bar(comparison["experiment"], comparison["best_valid_acc"])

plt.title("Experiment Comparison")
plt.ylabel("Validation Accuracy (%)")

path = fig_dir / "experiment_comparison.png"
plt.savefig(path, dpi=200)

plt.show()

print("Saved:", path)
```
